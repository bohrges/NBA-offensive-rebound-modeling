## **Stacking Ensemble**

#### **Imports and Config**

```python
import json
import os
import mlflow
import mlflow.sklearn
import numpy as np
import optuna
import pandas as pd
from optuna.integration.mlflow import MLflowCallback 
from lightgbm import LGBMClassifier
from sklearn.ensemble import RandomForestClassifier, StackingClassifier
from sklearn.linear_model import LogisticRegression 
from sklearn.inspection import permutation_importance
from sklearn.model_selection import cross_validate
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline
from xgboost import XGBClassifier
import matplotlib.pyplot as plt
import optuna.visualization.matplotlib as optuna_mpl

import warnings
warnings.filterwarnings("ignore", message="X does not have valid feature names", category=UserWarning)
optuna.logging.set_verbosity(optuna.logging.WARNING)
```

```python
# Run Configuration
MODEL_TYPE = "Stacking"
MOMENT = "moment_1"

N_FOLDS = 5
RANDOM_STATE = 42
N_TRIALS = 20 # fewer trials - stacking CV is expensive

TARGET_COL = "REBOUND_TYPE" # 1 = offensive, 0 = defensive
BALANCED   = False  # this file is the unbalanced version - see directory name
BASE_TAG   = "balanced" if BALANCED else "unbalanced"
SUFFIX     = f"_{BASE_TAG}"
DATA_PATH  = f"../../../../data/sequences/{MOMENT}/features/tabular/sequences_fe_tabular{SUFFIX}.csv"

GIT_COMMIT = "fd0a1ba"

# Overfitting control
# Composite objective: roc_auc_mean - ALPHA x (train_mean - test_mean)
ALPHA = 0.1
MAX_OVERFIT_GAP = 0.15

# Permutation importance (final model)
N_REPEATS_PERM = 10

# MLflow
EXP_BASELINE = f"{MODEL_TYPE}_baseline"
EXP_TUNED = f"{MODEL_TYPE}_tuned"
EXP_FINAL = f"{MODEL_TYPE}_final"
MLFLOW_DIR = os.path.abspath(os.path.join(os.path.dirname(os.path.abspath('__file__')), '../../../..', 'runs', 'mlruns'))
RUN_DIR = os.path.abspath(os.path.join(os.path.dirname(os.path.abspath('__file__')), '../../../..', 'runs', f"{MODEL_TYPE}_{MOMENT}_{BASE_TAG}"))

os.makedirs(MLFLOW_DIR, exist_ok=True)
os.makedirs(RUN_DIR, exist_ok=True)
os.makedirs(os.path.join(MLFLOW_DIR, ".trash"), exist_ok=True)
mlflow.set_tracking_uri(f"file:///{MLFLOW_DIR.replace(os.sep, '/')}")

print(f"Artifacts -> {RUN_DIR}")
print(f"Overfitting - alpha={ALPHA}, hard cap={MAX_OVERFIT_GAP}")
```

#### **Data Loading & Feature Selection**

```python
df = pd.read_csv(DATA_PATH)

EXCLUDED_COLS = ["GAME_ID", "EVENT_NUM", "DETECTION_CONFIDENCE",
                  "SHOT_DETECTION_CONFIDENCE", "RIM_DETECTION_CONFIDENCE"]
FEATURE_COLS = [c for c in df.columns if c != TARGET_COL and c not in EXCLUDED_COLS]

X = df[FEATURE_COLS].values
y = df[TARGET_COL].values
groups = df["GAME_ID"].values

print(f"Dataset : {X.shape[0]} samples | {X.shape[1]} features")
print(f"Class balance - offensive: {y.mean():.3f}")

# Class-balance flag, logged alongside GIT_COMMIT
class_counts = pd.Series(y).value_counts()
CLASS_BALANCE_RATIO = class_counts.min() / class_counts.max()
IS_BALANCED = bool(np.isclose(CLASS_BALANCE_RATIO, 1.0, atol=0.02))
```

#### **Load Base Learner Params from MLflow**

Pulls the best params logged in each model's `_final` experiment for this moment and balance,  
so the stack always uses the tuned configurations without manual copy-paste.

```python
# Cast a string MLflow param to its most likely Python type.
def cast_param(value):
    if value is None or value == "None":
        return None
    if value.lower() == "true":
        return True
    if value.lower() == "false":
        return False
    try:
        return int(value)
    except ValueError:
        pass
    try:
        return float(value)
    except ValueError:
        pass
    return value


def load_final_params(model_type, param_keys):
    exp_name = f"{model_type}_final"
    exp = mlflow.get_experiment_by_name(exp_name)
    if exp is None:
        raise RuntimeError(f"MLflow experiment '{exp_name}' not found. Run the {model_type} notebook first.")
    runs = mlflow.search_runs(
        experiment_ids=[exp.experiment_id],
        filter_string=(
            f"tags.mlflow.runName = '{model_type}' "
            f"and params.moment = '{MOMENT}' "
            f"and params.class_balanced = '{str(IS_BALANCED)}'"
        ),
    )
    if runs.empty:
        raise RuntimeError(
            f"No run '{model_type}' found in experiment '{exp_name}' for moment={MOMENT}, "
            f"class_balanced={IS_BALANCED}. Run the {model_type} notebook for this moment/balance first."
        )
    run = runs.sort_values("start_time", ascending=False).iloc[0]
    params = {
        k: cast_param(run[f"params.{k}"])
        for k in param_keys
        if f"params.{k}" in run.index and run[f"params.{k}"] is not None
    }
    print(f"  {model_type}: loaded {len(params)} params")
    return params


# Whitelisted param keys per model (meta-params like phase/model_type excluded)
# class_weight/scale_pos_weight keep the base learners' class weighting
RF_KEYS = ["n_estimators", "max_depth", "min_samples_leaf", "min_samples_split",
           "max_features", "bootstrap", "max_samples", "class_weight"]
XGB_KEYS = ["n_estimators", "max_depth", "learning_rate", "subsample",
            "colsample_bytree", "min_child_weight", "reg_alpha", "reg_lambda",
            "scale_pos_weight"]
LGBM_KEYS = ["n_estimators", "num_leaves", "learning_rate", "subsample",
             "subsample_freq", "colsample_bytree", "min_child_samples",
             "reg_alpha", "reg_lambda", "scale_pos_weight"]
LR_KEYS = ["C", "penalty", "solver", "max_iter", "l1_ratio", "class_weight"]

print("Loading base learner params from MLflow...")
rf_params = load_final_params("RandomForest", RF_KEYS)
xgb_params = load_final_params("XGBoost", XGB_KEYS)
lgbm_params = load_final_params("LightGBM", LGBM_KEYS)
lr_params = load_final_params("LogisticRegression", LR_KEYS)
print("Done.")
```

#### **Helpers**

```python
CV = StratifiedGroupKFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)
SCORING = {
    "accuracy": "accuracy",
    "roc_auc": "roc_auc",
    "f1": "f1_macro",
    "precision": "precision_macro",
    "recall": "recall_macro",
}

BASE_PARAMS = {
    "model_type": MODEL_TYPE,
    "moment": MOMENT,
    "git_commit": GIT_COMMIT,
    "n_features": len(FEATURE_COLS),
    "n_folds": N_FOLDS,
    "cv_strategy": "StratifiedGroupKFold",
    "random_state": RANDOM_STATE,
    "alpha": ALPHA,
    "max_overfit_gap": MAX_OVERFIT_GAP,
    "class_balanced": IS_BALANCED,
}


# Fit a MinMaxScaler PER FOLD (train only), then the model.
#
# Scaling happens inside CV so no validation-fold range leaks into training.
def wrap(model):
    return Pipeline([("scale", MinMaxScaler()), ("model", model)])


# Cross-validate and log metrics + overfitting diagnostics to MLflow.
#
# Returns a flat metrics dict that always includes:
#   - per-metric mean/std/train_mean for every SCORING key
#   - overfit_gap_<metric> = train_mean - test_mean
#   - composite_score = roc_auc_mean - alpha * overfit_gap_roc_auc
def run_cv_and_log(model, extra_params: dict, phase: str = None) -> dict:
    cv_results = cross_validate(
        wrap(model), X, y,
        groups=groups,
        cv=CV,
        scoring=SCORING,
        return_train_score=True,
        n_jobs=-1,
    )

    metrics = {}
    for metric in SCORING:
        test_scores  = cv_results[f"test_{metric}"]
        train_scores = cv_results[f"train_{metric}"]
        test_mean = float(np.mean(test_scores))
        train_mean = float(np.mean(train_scores))
        gap = train_mean - test_mean

        metrics[f"{metric}_mean"] = test_mean
        metrics[f"{metric}_std"] = float(np.std(test_scores))
        metrics[f"{metric}_train_mean"] = train_mean
        metrics[f"overfit_gap_{metric}"] = gap

    metrics["composite_score"] = (
        metrics["roc_auc_mean"] - ALPHA * metrics["overfit_gap_roc_auc"]
    )

    mlflow.log_params({**BASE_PARAMS, **extra_params})
    mlflow.log_metrics(metrics)

    if phase in ("baseline", "final"):
        # Feature list/order the saved model was trained on (read by the holdout cell)
        features_path = os.path.join(RUN_DIR, "features.json")
        with open(features_path, "w") as f:
            json.dump(FEATURE_COLS, f, indent=2)

        cv_path = os.path.join(RUN_DIR, f"cv_results_{phase}.json")
        with open(cv_path, "w") as f:
            json.dump({k: v.tolist() for k, v in cv_results.items()}, f, indent=2)

    return metrics


def print_summary(label: str, metrics: dict) -> None:
    gap = metrics["overfit_gap_roc_auc"]
    print(
        f"{label}\n"
        f"  roc_auc   : {metrics['roc_auc_mean']:.4f} +/- {metrics['roc_auc_std']:.4f}  "
        f"(train {metrics['roc_auc_train_mean']:.4f}, gap {gap:+.4f})\n"
        f"  accuracy  : {metrics['accuracy_mean']:.4f} +/- {metrics['accuracy_std']:.4f}  "
        f"(train {metrics['accuracy_train_mean']:.4f}, gap {metrics['overfit_gap_accuracy']:+.4f})\n"
        f"  composite : {metrics['composite_score']:.4f}  "
        f"[= roc_auc - {ALPHA}*gap]"
    )


# Construct the 4 tuned base learners from the params loaded above.
#
# Returned as plain (unwrapped) estimators: the outer wrap() pipeline scales the
# StackingClassifier's input once, so every base learner (and the meta LR) already
# sees scaled data - they do not need their own scaling step.
def build_base_estimators():
    return [
        ("rf", RandomForestClassifier(**rf_params, random_state=RANDOM_STATE, n_jobs=-1)),
        ("xgb", XGBClassifier(**xgb_params, random_state=RANDOM_STATE, n_jobs=-1)),
        ("lgbm", LGBMClassifier(**lgbm_params, random_state=RANDOM_STATE, n_jobs=-1)),
        ("lr", LogisticRegression(**lr_params, random_state=RANDOM_STATE, n_jobs=-1)),
    ]
```

#### **Baseline**

> **Note** - stacking CV is nested: the outer `cross_validate` (5 folds) wraps  
> `StackingClassifier`'s own inner CV (5 folds) for out-of-fold meta-feature generation.  
> This means 25 base model fits per base learner per outer fold, so expect longer  
> runtimes than the single-model notebooks.

```python
BASELINE_META_PARAMS = {
    "meta_C": 1.0,
    "meta_penalty": "l2",
    "passthrough": False,
}

mlflow.set_experiment(EXP_BASELINE)

with mlflow.start_run(run_name=MODEL_TYPE):
    model = StackingClassifier(
        estimators=build_base_estimators(),
        final_estimator=LogisticRegression(
            C=BASELINE_META_PARAMS["meta_C"],
            penalty=BASELINE_META_PARAMS["meta_penalty"],
            solver="lbfgs",
            max_iter=1000,
            random_state=RANDOM_STATE,
            # The meta-learner needs its own class weighting
            class_weight="balanced",
        ),  
        cv=5,
        passthrough=BASELINE_META_PARAMS["passthrough"],
        n_jobs=-1,
    )
    metrics = run_cv_and_log(
        model,
        extra_params={**BASELINE_META_PARAMS, "phase": "baseline"},
        phase="baseline",
    )

print_summary("Baseline", metrics)
```

#### **Tuned (Optuna)**

**Objective**: `composite_score = roc_auc_mean - alpha x overfit_gap`  
Trials where `overfit_gap_roc_auc > MAX_OVERFIT_GAP` are penalized to 0.0.

Base learner params are fixed (already tuned). Optuna searches over the meta-learner  
and the `passthrough` flag, which controls whether the original features are also passed  
to the meta-learner alongside the base learner predictions.

```python
mlflow.set_experiment(EXP_TUNED)

mlflc = MLflowCallback(
    tracking_uri=mlflow.get_tracking_uri(),
    metric_name="composite_score",
    create_experiment=False,
)

def progress_callback(study, trial):
    print(f"Trial {trial.number + 1}/{N_TRIALS} - composite: {trial.value:.4f} | best so far: {study.best_value:.4f}")


@mlflc.track_in_mlflow()
def objective(trial):
    penalty = trial.suggest_categorical("meta_penalty", ["l1", "l2"])
    # l1 requires liblinear; l2 works with lbfgs (faster for larger datasets)
    solver = "liblinear" if penalty == "l1" else "lbfgs"

    params = {
        "meta_C": trial.suggest_float("meta_C", 1e-3, 1e2, log=True),
        "meta_penalty": penalty,
        "meta_solver": solver,
        "passthrough": trial.suggest_categorical("passthrough", [True, False]),
    }

    model = StackingClassifier(
        estimators=build_base_estimators(),
        final_estimator=LogisticRegression(
            C=params["meta_C"],
            penalty=params["meta_penalty"],
            solver=solver,
            max_iter=1000,
            random_state=RANDOM_STATE,
            class_weight="balanced",
        ),
        cv=5,
        passthrough=params["passthrough"],
        n_jobs=-1,
    )

    metrics = run_cv_and_log(model, extra_params={**params, "phase": "tuned"})

    gap = metrics["overfit_gap_roc_auc"]

    if gap > MAX_OVERFIT_GAP:
        mlflow.log_metric("pruned_overfit", 1.0)
        return 0.0

    mlflow.log_metric("pruned_overfit", 0.0)
    return metrics["composite_score"]


study = optuna.create_study(
    direction="maximize",
    study_name=f"{MODEL_TYPE}_{MOMENT}_{BASE_TAG}",
    sampler=optuna.samplers.TPESampler(seed=RANDOM_STATE),
)
study.optimize(objective, n_trials=N_TRIALS, callbacks=[mlflc, progress_callback])

print(f"\nBest trial - composite_score: {study.best_value:.4f}")
print(f"Best params: {study.best_params}")
```

```python
# Inspect overfitting
trials_df = study.trials_dataframe()

valid = trials_df[trials_df["value"] > 0].copy()

print(f"Valid trials (gap <= {MAX_OVERFIT_GAP}): {len(valid)} / {len(trials_df)}")
print(valid[["value"]].describe().T.to_string())
```

```python
with mlflow.start_run(run_name=f"best_{MODEL_TYPE}"):
    mlflow.log_params({
        **BASE_PARAMS,
        **study.best_params,
        "phase": "tuned_best",
        "source": "optuna_best",
    })
    mlflow.log_metric("composite_score", study.best_value)

    optuna_mpl.plot_param_importances(study)
    plt.tight_layout()
    plt.savefig(os.path.join(RUN_DIR, "optuna_param_importance.png"), dpi=150)
    plt.close()
```

#### **Final**

```python
best = study.best_params
final_solver = "liblinear" if best["meta_penalty"] == "l1" else "lbfgs"

mlflow.set_experiment(EXP_FINAL)

with mlflow.start_run(run_name=MODEL_TYPE) as final_run:
    model = StackingClassifier(
        estimators=build_base_estimators(),
        final_estimator=LogisticRegression(
            C=best["meta_C"],
            penalty=best["meta_penalty"],
            solver=final_solver,
            max_iter=1000,
            random_state=RANDOM_STATE,
            class_weight="balanced",
        ),
        cv=5,
        passthrough=best["passthrough"],
        n_jobs=-1,
    )
    metrics = run_cv_and_log(
        model,
        extra_params={**best, "meta_solver": final_solver, "phase": "final"},
        phase="final",
    )

    # Refit the full pipeline (scaler + stack) on all data and log it as the artifact.
    # (Without the pipeline, this would train the LR base and passthrough meta on RAW data.)
    final_pipe = wrap(model)
    final_pipe.fit(X, y)
    mlflow.sklearn.log_model(final_pipe, "model")

FINAL_RUN_ID = final_run.info.run_id
print_summary("Final", metrics)
```

```python
from sklearn.model_selection import cross_val_predict

# Out-of-fold probabilities, used by calibrate.py
oof_probs = cross_val_predict(
    wrap(model), X, y,
    groups=groups,
    cv=CV,
    method="predict_proba",
    n_jobs=-1,
)[:, 1]  # P(offensive)

oof_df = pd.DataFrame({
    "GAME_ID": df["GAME_ID"].values,
    "EVENT_NUM": df["EVENT_NUM"].values,
    "y_true": y,
    "y_prob": oof_probs,
})

oof_path = os.path.join(RUN_DIR, "oof_predictions.csv")
oof_df.to_csv(oof_path, index=False)

with mlflow.start_run(run_id=FINAL_RUN_ID):
    mlflow.log_artifact(oof_path)

print(f"OOF predictions: {len(oof_df)} rows | true offensive rate {oof_df['y_true'].mean():.3f} | mean predicted prob {oof_df['y_prob'].mean():.3f}")
```

#### **Feature Importance (Final Model)**

`StackingClassifier` has no native feature importances. Permutation importance is used  
on the last CV fold's validation split, measuring the drop in ROC-AUC when each input  
feature is shuffled. This reflects the full stack's sensitivity, including the meta-learner.

```python
train_idx, val_idx = list(CV.split(X, y, groups))[-1]
X_val, y_val = X[val_idx], y[val_idx]

perm = permutation_importance(
    final_pipe, X_val, y_val,
    n_repeats=N_REPEATS_PERM,
    scoring="roc_auc",
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

importances = pd.Series(perm.importances_mean, index=FEATURE_COLS)
top20 = importances.nlargest(20)

fig, ax = plt.subplots(figsize=(8, 6))
top20.sort_values().plot.barh(ax=ax)
ax.set_title("Stacking Feature Importances - Top 20 (Permutation)")
ax.set_xlabel("Mean decrease in ROC-AUC")
plt.tight_layout()

fig_path = os.path.join(RUN_DIR, "feature_importances.png")
fig.savefig(fig_path, dpi=150)
plt.show()
print(f"Saved -> {fig_path}")
```

#### **Holdout Evaluation**

```python
from sklearn.metrics import accuracy_score, roc_auc_score, f1_score, precision_score, recall_score

# Holdout counterpart of DATA_PATH (holdout is never balanced)
HOLDOUT_DATA_PATH = f"../../../../data/sequences/{MOMENT}/features/tabular/sequences_fe_tabular_holdout.csv"

df_holdout = pd.read_csv(HOLDOUT_DATA_PATH)

with open(os.path.join(RUN_DIR, "features.json")) as f:
    holdout_feature_cols = json.load(f)

missing = set(holdout_feature_cols) - set(df_holdout.columns)
assert not missing, f"Holdout is missing features used in training: {missing}"

X_holdout = df_holdout[holdout_feature_cols].values
y_holdout = df_holdout[TARGET_COL].values

y_pred = final_pipe.predict(X_holdout)
y_score = (final_pipe.predict_proba(X_holdout)[:, 1] if hasattr(final_pipe, "predict_proba")
           else final_pipe.decision_function(X_holdout))

holdout_metrics = {
    "accuracy": accuracy_score(y_holdout, y_pred),
    "roc_auc": roc_auc_score(y_holdout, y_score),
    "f1": f1_score(y_holdout, y_pred, average="macro"),
    "precision": precision_score(y_holdout, y_pred, average="macro"),
    "recall": recall_score(y_holdout, y_pred, average="macro"),
}

with mlflow.start_run(run_id=FINAL_RUN_ID):
    mlflow.log_metrics({f"holdout_{k}": v for k, v in holdout_metrics.items()})
    mlflow.log_param("n_holdout", len(df_holdout))

print(f"Holdout: {len(df_holdout)} rows, offensive rate {y_holdout.mean():.3f}")
for k, v in holdout_metrics.items():
    print(f"  holdout_{k:9s}: {v:.4f}")
```
